```python
import duckdb
import pandas as pd

con = duckdb.connect(
    "../data/reep-register-v1.duckdb",
    read_only=True
)

print(con.execute("SHOW TABLES").fetchdf())
```

```python
table_list = con.execute("SHOW TABLES").fetchdf()['name']
for table in table_list:
  df = con.execute("SELECT * FROM " + table).fetchdf()
  headers = df.columns
  print(table)
  print(headers)
  print()
```
