```python
from bs4 import BeautifulSoup
import requests
```

```python
player_id = 148455
url = f'https://tmapi.transfermarkt.technology/player/{player_id}/market-value-history'

page = requests.get(url)

print(page.status_code)
print(page.url)

data = page.json()
print(data)
```
